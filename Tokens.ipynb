# Types of tokens
- bigrams - tokens with 2 words
- trigrams - tokens with 3 words
- ngrams - token with more than 3 words 

```python
from nltk.util import bigrams, trigrams, ngrams
bigrams = list(bigrams(word_tokens))
bigrams
```

```python
from nltk.util import bigrams, trigrams, ngrams
trigrams = list(trigrams(word_tokens))
trigrams
```

```python
from nltk.util import bigrams, trigrams, ngrams
ngrams = list(ngrams(word_tokens))
ngrams
```

```python
from nltk.util import bigrams, trigrams, ngrams
ngrams = list(ngrams(word_tokens, 5))
ngrams
```

```python
#  What is a token? Tokenize the sentence and count the total tokens. I love learning Artificial Intelligence
#  A token is the smallest unit of text processed by an NLP or LLM model.

text = "I love learning Artificial Intelligence"
tokens = text.split()
print(tokens)
print("Token Count:", len(tokens))
```

```python
#     Print every token with its index.

text = "I love learning NLP"
tokens = text.split()
for i, token in enumerate(tokens):
print(i, token)                    # indentation error
```

```python
#   Print every token with its index.

text = "I love learning NLP"
tokens = text.split()
for i, token in enumerate(tokens):
    print(i, token)    
```

```python
#  Find the longest token in the sentence. Machine Learning is amazing

text = "Machine Learning is amazing"
tokens = text.split()
longest = max(tokens, key=len)
print(longest)
```

```python
#  Find the shortest token.

text = "Machine Learning is amazing"
tokens = text.split()
shortest = min(tokens, key=len)
print(shortest)
```

```python
#   Count frequency of every token.

from collections import Counter
text = "AI AI AI NLP NLP"
tokens = text.split()
freq = Counter(tokens)
print(freq)
```

```python
#  Find unique tokens

text = "AI AI NLP NLP DeepLearning"
tokens = text.split()
unique_tokens = set(tokens)
print(unique_tokens)
```

```python
#    Create vocabulary from text.

text = "I love AI and NLP"
tokens = text.split()
vocab = sorted(set(tokens))
print(vocab)

```

```python
#     Convert tokens into token IDs.

text = "I love AI"
tokens = text.split()
vocab = {word:i for i,word in enumerate(set(tokens))}
token_ids = [vocab[word] for word in tokens]
print(token_ids)

```

```python
#  Decode token IDs back into text.

vocab = {
"I":0,
"love":1,
"AI":2
}
reverse_vocab = {v:k for k,v in vocab.items()}
ids = [0,1,2]
sentence = [reverse_vocab[i] for i in ids]
print(" ".join(sentence))
```

```python
#   Implement a simple tokenizer class.

class SimpleTokenizer:
    def tokenize(self,text):
        return text.split()
obj = SimpleTokenizer()
print(obj.tokenize("I love AI"))
```
